```python
import tensorflow as tf

gpus = tf.config.list_physical_devices('GPU')

if gpus:
    for gpu in gpus:
        details = tf.config.experimental.get_device_details(gpu)
        print(details.get('device_name', 'Unknown GPU'))
else:
    print("No GPU found")
```
