```python
from ultralytics import YOLO
import cv2
import time

# Load video
video_path = "/home/saidul/Desktop/4-2/Deep Learning/data/Object_detection_video/computer_lab_18sec.mp4" 

# Models to compare
models = {
    "YOLOv8n": "yolov8n.pt",
    "YOLOv11n": "yolo11n.pt",
    "YOLOv12n": "yolo12n.pt" 
}

results_summary = {}

for model_name, weight in models.items():
    print(f"\n Running {model_name}...")
    model = YOLO(weight)
    
    start_time = time.time()
    results = model.predict(
        source=video_path,
        save=True,           # Saves output video
        conf=0.25,
        iou=0.45,
        stream=True,
        verbose=False
    )
    end_time = time.time()
    
    total_time = end_time - start_time

    frame_count = 0
    for _ in results:
        frame_count += 1
    fps = frame_count / total_time if total_time > 0 else 0

    print(f"{model_name} finished in {total_time:.2f} seconds | Approx FPS: {fps:.1f}")
    results_summary[model_name] = {"time": total_time, "fps": fps}

print("\n=== Comparison Summary ===")
for name, data in results_summary.items():
    print(f"{name:10s} → Time: {data['time']:.2f}s | FPS: {data['fps']:.1f}")
```
