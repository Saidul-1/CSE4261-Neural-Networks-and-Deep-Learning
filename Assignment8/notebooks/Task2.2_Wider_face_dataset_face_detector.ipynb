```python
import os
import shutil
import random

# Set paths
source = "/home/saidul/Desktop/4-2/Deep Learning/data/wider_face/archive"   # Change if needed
base = "/home/saidul/Desktop/4-2/Deep Learning/data/wider_face/wider_face_small"

# Create small subset (Recommended: 800-1200 images for CPU)
os.makedirs(f"{base}/images/train", exist_ok=True)
os.makedirs(f"{base}/images/val", exist_ok=True)
os.makedirs(f"{base}/labels/train", exist_ok=True)
os.makedirs(f"{base}/labels/val", exist_ok=True)

# Take only 1000 images for train, 200 for val
image_list = [f for f in os.listdir(f"{source}/images") if f.endswith('.jpg')]
random.shuffle(image_list)

train_imgs = image_list[:100]
val_imgs = image_list[100:120]

def copy_files(img_list, split):
    for img in img_list:
        label = img.replace('.jpg', '.txt')
        shutil.copy(f"{source}/images/{img}", f"{base}/images/{split}/{img}")
        if os.path.exists(f"{source}/labels/{label}"):
            shutil.copy(f"{source}/labels/{label}", f"{base}/labels/{split}/{label}")

copy_files(train_imgs, "train")
copy_files(val_imgs, "val")

print("Small dataset created successfully!")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")   # Nano model - best for CPU

results = model.train(
    data="/home/saidul/Desktop/4-2/Deep Learning/data/wider_face/data.yaml",
    epochs=5,              # Keep low for CPU
    imgsz=640,
    batch=4,                # Very important for your RAM
    name="yolov8_face_cpu",
    patience=10,
    device="cpu",
    augment=True,
    workers=2
)
```
