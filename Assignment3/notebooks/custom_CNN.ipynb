```python

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dropout, Flatten, Dense, BatchNormalization

def create_custom_cnn(activation='relu'):
    model = Sequential([
        Conv2D(32, (3,3), padding='same', activation=activation, input_shape=(32,32,3)),
        BatchNormalization(),
        Conv2D(32, (3,3), padding='same', activation=activation),
        BatchNormalization(),
        MaxPooling2D((2,2)),
        Dropout(0.25),
        
        Conv2D(64, (3,3), padding='same', activation=activation),
        BatchNormalization(),
        Conv2D(64, (3,3), padding='same', activation=activation),
        BatchNormalization(),
        MaxPooling2D((2,2)),
        Dropout(0.25),
        
        Conv2D(128, (3,3), padding='same', activation=activation),
        BatchNormalization(),
        MaxPooling2D((2,2)),
        Dropout(0.25),
        
        Flatten(),
        Dense(256, activation=activation),
        BatchNormalization(),
        Dropout(0.5),
        Dense(10, activation='softmax')
    ])
    return model
```

```python
import tensorflow as tf
from tensorflow.keras.datasets import cifar10

# 1. Load and Preprocess CIFAR-10 Data
(x_train, y_train), (x_test, y_test) = cifar10.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0  # Scale pixels to [0, 1]

# 2. Define the list of activations to test
activations = ['relu', 'sigmoid', 'softmax']
histories = {}

for act in activations:
    # Create a fresh instance of your model with the specific activation
    model = create_custom_cnn(activation=act)
    
    model.compile(optimizer='adam', 
                  loss='sparse_categorical_crossentropy', 
                  metrics=['accuracy'])
    
    print(f"\n--- Training with {act} activation ---")
    
    # Train (reduced epochs to 10 for faster testing, change back to 30 if needed)
    history = model.fit(x_train, y_train, 
                        epochs=3, 
                        batch_size=64, 
                        validation_data=(x_test, y_test), 
                        verbose=1)
    
    histories[act] = history
```

```python
# Evaluate
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Test Accuracy: {test_acc:.4f}")
```
